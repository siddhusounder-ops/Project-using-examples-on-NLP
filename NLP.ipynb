### Chunking
Chunking uses a special regexp syntax for rules that delimit the chunks. These rules must be converted to 'regular' regular expressions before a sentence can be chunked.


```python
from nltk.chunk import *
from nltk.chunk.util import *
from nltk.chunk.regexp import *
from nltk import Tree

tagged_text = "[ The/DT cat/NN ] sat/VBD on/IN [ the/DT mat/NN ] [ the/DT dog/NN ] chewed/VBD ./."
gold_chunked_text = tagstr2tree(tagged_text)
unchunked_text = gold_chunked_text.flatten()
print (unchunked_text)

chunk_rule = ChunkRule("<.*>+", "Chunk everything")
chink_rule = ChunkRule("<VBD|IN|\.>", "Chunk on verbs/prepositions")

chunk_parser = RegexpChunkParser([chunk_rule], chunk_label='NP')
chunked_text = chunk_parser.parse(unchunked_text)
print(chunked_text)
print(chunk_parser)
print(repr(chunk_parser))
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import RegexpParser

# Sample text for demonstration
text = "The quick brown fox jumps over the lazy dog."

# Tokenize and POS-tag the text
tokens = word_tokenize(text)
tagged_words = pos_tag(tokens)
print("POS Tagged Words:", tagged_words)

# --- 1. Noun Phrase (NP) Chunking ---
# Grammar for Noun Phrase: optional Determiner (DT), any number of Adjectives (JJ), followed by a Noun (NN).
np_grammar = "NP: {<DT>?<JJ>*<NN>}"
np_chunk_parser = RegexpParser(np_grammar)
np_chunked = np_chunk_parser.parse(tagged_words)
print("\n1. Noun Phrase (NP) Chunking:")
print(np_chunked)

# --- 2. Verb Phrase (VP) Chunking ---
# Grammar for Verb Phrase: any number of verbs (VB, VBD, VBG, VBN, VBP, VBZ).
vp_grammar = "VP: {<VB.?><PRP|NNP>?}" # Simple VP: Verb followed by optional pronoun or proper noun
vp_chunk_parser = RegexpParser(vp_grammar)
vp_chunked = vp_chunk_parser.parse(tagged_words)
print("\n2. Verb Phrase (VP) Chunking:")
print(vp_chunked)

# --- 3. Chunking based on specific word patterns (e.g., Adjective-Noun pairs) ---
# This is a variation of NP chunking, but specifically targeting JJ+NN
adj_nn_grammar = "ADJ_NN: {<JJ><NN>}"
adj_nn_chunk_parser = RegexpParser(adj_nn_grammar)
adj_nn_chunked = adj_nn_chunk_parser.parse(tagged_words)
print("\n3. Adjective-Noun Chunking:")
print(adj_nn_chunked)

# --- 4. Chinking (removing parts from a chunk) ---
# Chinking allows you to define what *not* to include in a chunk.
# Here, we'll chunk everything, but then chink out prepositions (IN).
# The rule specifies to chunk everything (<.*>+) except for the pattern defined in the chink.
# Corrected: Added a 'CHUNK:' label before the pattern.
chink_grammar = "CHUNK: {<.*>+}<IN>+"
chink_parser = RegexpParser(chink_grammar)
chinked_text = chink_parser.parse(tagged_words)
print("\n4. Chinking (removing prepositions):")
print(chinked_text)

# --- Additional example: Word-level chunking for specific length (conceptual, usually handled by tokenization/filtering) ---
# While NLTK's chunking is primarily rule-based on POS tags, you can achieve 'length' based filtering post-chunking
# or by making very specific regex for word types that match certain length patterns.
# For example, to 'chunk' words of a specific length, it's more about filtering tokens.
print("\n5. Conceptual 'Length' based filtering (not direct NLTK chunking but related):")
long_words = [word for word, tag in tagged_words if len(word) > 4]
print(f"Words longer than 4 characters: {long_words}")

# 'Line' and 'Page' chunking are generally handled by splitting the raw text first,
# before tokenization and POS tagging, as NLTK's chunking operates on tagged word sequences.
# Example of line-based split:
multi_line_text = "First line of text.\nSecond line here.\nAnd a third one."
lines = multi_line_text.split('\n')
print("\n6. Line-based Segmentation (pre-chunking step):")
for i, line in enumerate(lines):
    print(f"Line {i+1}: {line}")
```

```python
def extract_entity_names(t):
    entity_names = []

    if hasattr(t, 'label') and t.label:
        if t.label() == 'NE':
            entity_names.append(' '.join([child[0] for child in t]))
        else:
            for child in t:
                entity_names.extend(extract_entity_names(child))

    return entity_names

entity_names = []
for tree in chunked_sentences:
    # Print results per sentence
    print (extract_entity_names(tree))
    print (entity_names.extend(extract_entity_names(tree)))

# Print all entity names
#print (entity_names)

# Print unique entity names
print (set(entity_names))
```

```python
!pip install gensim
import gensim.downloader as api
import numpy as np

# Load a pre-trained Word2Vec model (this might take a while the first time)
# You can choose other models from `api.info()['models']`
print("Downloading Word2Vec model...")
word_vectors = api.load("word2vec-google-news-300")
print("Word2Vec model loaded successfully.")

def sentence_to_vector(sentence, model):
    words = sentence.lower().split()
    # Filter out words not in the model's vocabulary
    vectorized_words = [model[word] for word in words if word in model.key_to_index]

    if not vectorized_words:
        return np.zeros(model.vector_size)  # Return a zero vector if no words are found

    # Average the word vectors to get the sentence vector
    return np.mean(vectorized_words, axis=0)

# Example sentence
sentence = "This is a great day for natural language processing. The learners are able to practice fundamentals of Gen AI."

sentence_vec = sentence_to_vector(sentence, word_vectors)

print(f"\nOriginal Sentence: {sentence}")
print(f"Sentence Vector (first 10 dimensions): {sentence_vec[:10]}")
print(f"Shape of Sentence Vector: {sentence_vec.shape}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Example sentences
sentence1 = "lazy dog."
sentence2 = "We are in NLP and word embedding session."
sentence3 = "learning similarity in word2vec."

# Convert sentences to vectors using the previously defined function
vector1 = sentence_to_vector(sentence1, word_vectors)
vector2 = sentence_to_vector(sentence2, word_vectors)
vector3 = sentence_to_vector(sentence3, word_vectors)

# Reshape vectors for cosine_similarity function if they are 1D
# cosine_similarity expects 2D arrays, e.g., [[vector]]
vector1_reshaped = vector1.reshape(1, -1)
vector2_reshaped = vector2.reshape(1, -1)
vector3_reshaped = vector3.reshape(1, -1)

# Calculate cosine similarity
similarity1_2 = cosine_similarity(vector1_reshaped, vector2_reshaped)[0][0]
similarity2_3 = cosine_similarity(vector2_reshaped, vector3_reshaped)[0][0]

print(f"\nSentence 1: {sentence1}")
print(f"Sentence 2: {sentence2}")
print(f"Sentence 3: {sentence3}")
print(f"\nCosine Similarity between Sentence 1 and Sentence 2: {similarity1_2:.4f}")
print(f"Cosine Similarity between Sentence 2 and Sentence 3: {similarity2_3:.4f}")
```
